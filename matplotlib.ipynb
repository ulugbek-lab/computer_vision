```python
import matplotlib.pyplot as plt
import numpy as np
```

# Basic Line graph

```python
x = np.array([1,2,3,4])
y = np.array([10,14,15,16])
plt.plot(x,y)
plt.title("Line Plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

# Line Styles and Colors

```python
from matplotlib.lines import lineStyles
x = np.array([1,2,3,15])
y = np.array([3,4,2,9])

plt.plot(y,linestyle = 'dotted', color= 'green', marker = 'o')
plt.plot(x,linestyle = 'dotted', color= 'blue', marker = 'o')
plt.title("Styled Line Plot")
plt.show()
```

# Multiple Lines in one graph with Numpy and Matplotlib

```python
x= np.array([1,3,5,6,])
y1= np.array([7,5,8,12,])
y2= np.array([1,2,4,9,])

plt.plot(x, y1, label= 'Line 1', color= 'red', marker= 's')
plt.plot(x, y2, label= 'Line 2', color= 'green', linestyle= '--', marker= 'o')
plt.title('Multiple Line Plot')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.legend()
plt.show()
```

## Bar Chart with Numpy and **Matplotlib**

```python
x = np.array(['A', 'B', 'C', 'D'])
y = np.array([3, 8, 1, 10])

plt.bar(x, y, color='blueviolet')
plt.title('Bar Chart')
plt.xlabel('Categories')
plt.ylabel('Numbers')
plt.show()
```

# Scatter Plot with Numpy and Matplotlib

```python
x = np.array([5,7,8,7,2])
y = np.array([1,5,2,3,7])

plt.scatter(x,y, color='orange', s=100)
plt.title("Scatter Plot Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

# Histogram with Numpy and Matplotlib

```python
data = np.array([3,5,6,7,9,8,8,10])
plt.hist(data, bins=4, color='orange', edgecolor='red')
plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

# Subplot(Multiple graphs Together)

```python
x = np.array([1,2,3,4])
y1 = np.array([10,14,15,16])
y2 = np.array([3,4,2,9])

plt.figure(figsize=(10,5)) # width, height


plt.subplot(1,2,1)
plt.plot(x,y1, color="red")
plt.title("Line Plot 1")

plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)

plt.subplot(1,2,2)
plt.bar(x,y2, color="blue")
plt.title("Bar Plot 2")

plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.tight_layout()
plt.show()
plt.savefig("subplot.png")
#
```
